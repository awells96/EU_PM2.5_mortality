# Sum country level mortality for all ensemble members

Using GBD adjusted country masks, total mortality for each country is calculated

```python
import os
import xarray as xr
import config
from utils.utils import require_dir
import pathlib
```

```python
# === Path config ===
MASK_DIR = require_dir(pathlib.Path(config.WORK_ROOT) / "BMR" / "masks" / "country")

in_file = "GBD_Country_Masks_EU_SHERPA_res.nc"
in_path = os.path.join(MASK_DIR, in_file)
country_mask = xr.open_dataarray(in_path)
```

```python
# === Scenario and path config ===
# For RR curves and file name
GBD_version = "GBD23"

scenarios = ["H", "HL", "L", "LN", "M", "ML", "VL"]
years = [2040, 2060, 2080, 2100]

MORTALITY_DIR = require_dir(pathlib.Path(config.WORK_ROOT) / "mortality" / "gridpoint_mortality")

for scenario in scenarios:
    for year in years:
        print(f"Processing {scenario}, Year {year}")
        in_file = f"Mortality_{GBD_version}_{scenario}_{year}.nc"
        in_path = os.path.join(MORTALITY_DIR, in_file)

        if not os.path.exists(in_path):
            print(f"Missing: {in_path}")
            continue

        da = xr.open_dataarray(in_path)

        # Vectorized over country
        mortality_country = (da * country_mask).sum(dim=("latitude", "longitude"), skipna=True)

        mortality_country.attrs["description"] = (
            "Country level mean Mortality due to PM2.5 - scripts by A.F. Wells (2025)"
        )
        mortality_country.attrs["scenario"] = scenario
        mortality_country.attrs["year"] = year
        mortality_country.attrs["GBD version"] = GBD_version

        out_file = f"Mortality_Country_sum_{GBD_version}_{scenario}_{year}.nc"
        out_path = os.path.join(MORTALITY_DIR, out_file)

        print(f"Saving country mortality to {out_path}")
        mortality_country.to_netcdf(out_path)

print("All processing complete.")
```
